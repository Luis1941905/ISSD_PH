3. Chatbot utilizando TF-IDF y Similitud del Coseno
Para construir el primer chatbot se utilizará la técnica TF-IDF (Term Frequency - Inverse Document Frequency).

TF-IDF transforma cada pregunta en un vector numérico considerando la importancia de las palabras dentro del conjunto de preguntas.

Posteriormente se utilizará la similitud del coseno para comparar la consulta del usuario con todas las preguntas almacenadas y seleccionar la más parecida.

La respuesta asociada a la pregunta más similar será devuelta por el chatbot.

```python
# Librerías necesarias

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

Preparación de los datos
Se separan las preguntas y respuestas del dataset para construir la matriz TF-IDF.

```python
preguntas = df["pregunta"].tolist()
respuestas = df["respuesta"].tolist()

print("Cantidad de preguntas:", len(preguntas))
```

Creación de la matriz TF-IDF
Cada pregunta será representada mediante un vector numérico.

```python
vectorizador = TfidfVectorizer()

matriz_tfidf = vectorizador.fit_transform(preguntas)

print("Dimensión de la matriz:", matriz_tfidf.shape)
```

Función del chatbot TF-IDF
La función recibe una consulta, calcula la similitud con todas las preguntas y devuelve la respuesta asociada a la pregunta más similar.

```python
def chatbot_tfidf(consulta):

    # Convertir la consulta en vector TF-IDF
    consulta_vector = vectorizador.transform([consulta])

    # Calcular similitudes
    similitudes = cosine_similarity(
        consulta_vector,
        matriz_tfidf
    )

    # Obtener índice de la pregunta más parecida
    indice = similitudes.argmax()

    # Retornar respuesta asociada
    return respuestas[indice]
```

Prueba del chatbot TF-IDF

```python
consulta = "¿Cómo puedo anotarme para rendir un examen final?"

respuesta = chatbot_tfidf(consulta)

print("Consulta:")
print(consulta)

print("\nRespuesta:")
print(respuesta)
```

Explicación de funcionamiento


Cuando el usuario realiza una consulta, el texto es transformado en un vector TF-IDF.

A continuación se calcula la similitud del coseno entre la consulta y todas las preguntas almacenadas.

La pregunta con mayor similitud es considerada la más cercana al significado de la consulta y se devuelve la respuesta correspondiente.

 Chatbot utilizando Embeddings, que es donde normalmente se obtienen los puntos más interesantes del TP.

Chatbot utilizando Embeddings


Los embeddings son representaciones vectoriales que permiten capturar el significado semántico de las palabras y frases.

A diferencia de TF-IDF, que trabaja principalmente con la frecuencia de las palabras, los embeddings permiten reconocer similitudes de significado incluso cuando se utilizan términos diferentes.

Para este trabajo se utilizará el modelo preentrenado es_core_news_md de SpaCy.

Justificación del embedding elegido


Se seleccionó el modelo preentrenado es_core_news_md de SpaCy porque:

Está entrenado para idioma español.
Incluye vectores semánticos (word embeddings).
Permite comparar significados además de palabras.
Reconoce similitudes entre términos relacionados, por ejemplo:
profesor ↔ docente
notas ↔ calificaciones
inscripción ↔ registro
Esto permite mejorar la recuperación de respuestas cuando el usuario formula una pregunta diferente a la almacenada en el dataset.

```python
# Instalación de SpaCy

!pip install spacy -q
!python -m spacy download es_core_news_md
```

```python
import spacy

nlp = spacy.load("es_core_news_md")

print("Modelo cargado correctamente")
```

Generación de embeddings para las preguntas


Cada pregunta del dataset será convertida a un vector semántico.

```python
embeddings = []

for pregunta in preguntas:

    vector = nlp(pregunta).vector

    embeddings.append(vector)

print("Cantidad de embeddings generados:", len(embeddings))
```

Construcción del chatbot basado en embeddings


La consulta del usuario también será convertida en un vector y luego se calculará la similitud del coseno con todas las preguntas almacenadas.

La respuesta asociada a la pregunta más similar será devuelta por el chatbot.

```python
def chatbot_embeddings(consulta):

    consulta_vector = nlp(consulta).vector.reshape(1, -1)

    similitudes = cosine_similarity(
        consulta_vector,
        embeddings
    )[0]

    indice = similitudes.argmax()

    return respuestas[indice]
```

Prueba del chatbot basado en embeddings

```python
consulta = "¿Cómo puedo hablar con un docente?"

respuesta = chatbot_embeddings(consulta)

print("Consulta:")
print(consulta)

print("\nRespuesta:")
print(respuesta)
```

Explicación de funcionamiento


En este enfoque cada pregunta es representada mediante un vector semántico generado por SpaCy.

Cuando el usuario realiza una consulta, también se genera un vector para dicha consulta.

Posteriormente se calcula la similitud del coseno entre el vector de la consulta y los vectores de todas las preguntas almacenadas.

La pregunta con mayor similitud es seleccionada y se devuelve la respuesta correspondiente.
